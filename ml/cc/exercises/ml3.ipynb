## Predicting the prices of the goldstocks

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
df = pd.read_csv('gold_stock.csv')
df
```

```python
df = df.drop(index=1)
df
```

```python
df.describe()
```

```python
df = df.rename(columns={
    'Price': 'Date',
    'Close': 'Closing_Price',
    'High': 'High_Price',
    'Low': 'Low_Price',
    'Open': 'Opening_Price'
})
df
```

```python
df_clean = df.drop(columns = ['Date','Closing_Price'])
x = df['Date']
y = df['Closing_Price']
df_clean
```

```python
df_clean.columns
```

```python
sns.heatmap(df_clean.corr(), annot = True, cmap = 'coolwarm')
```

```python
scalar = StandardScaler()

columns = ['Closing_Price', 'High_Price', 'Low_Price', 'Opening_Price', 'Volume']
df_clean[columns] = scalar.fit_transform(df_clean[columns])
df_clean.head()

```

```python
x_train, x_test, y_train, y_test = train_test_split(df_clean, y, test_size = 0.20, random_state = 42)

# model = LinearRegression()
# model.fit
```

```python
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
y_pred
```

```python
y_test
```

```python
# r2 score
r2 = r2_score(y_test, y_pred)
r2
```
